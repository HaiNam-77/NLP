```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt_tab')
nltk.download('wordnet')
text = "The bats are hanging on their feet for best performance. The wolves are howling loudly!"

tokens = word_tokenize(text)

stemming = PorterStemmer()
stemming_words = [stemming.stem(word) for word in tokens]

lemmatization = WordNetLemmatizer()
lemmatization_words = [lemmatization.lemmatize(word) for word in tokens]

print("Original Tokens: ", tokens)
print("-" * 50)
print("Stemming Result: ", stemming_words)
# Lưu ý: 'wolves' -> 'wolv', 'hanging' -> 'hang'
print("-" * 50)
print("Lemmatization Result: ", lemmatization_words)
```
