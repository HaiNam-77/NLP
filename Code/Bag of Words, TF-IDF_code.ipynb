```python
import pandas as pd
import math
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# ==========================================
# 0. DỮ LIỆU ĐẦU VÀO
# ==========================================
corpus = [
    "code python rất vui và code java rất khó",
    "sinh viên học code python",
    "chuyên gia viết phần mềm"
]
# Khung từ vựng chuẩn do con người định nghĩa
vocab = ["code", "python", "rất", "vui", "và", "java", "khó", "sinh", "viên", "học", "chuyên", "gia", "viết", "phần", "mềm"]
N = len(corpus) # Tổng số câu (N=3)


# ==========================================
# PHẦN 1: TỰ CODE PYTHON THUẦN (MANUAL BoW)
# ==========================================
manual_bow = []
for sentence in corpus:
    words = sentence.split()
    vector = [words.count(w) for w in vocab]
    manual_bow.append(vector)

print("--- 1. MA TRẬN ĐẶC (DENSE) TỪ CODE THUẦN (BoW) ---")
print(pd.DataFrame(manual_bow, columns=vocab))
print("\n")


# ==========================================
# PHẦN 2: DÙNG THƯ VIỆN SKLEARN (BoW)
# ==========================================
print("--- 2. MA TRẬN THƯA (SPARSE) TỪ SKLEARN (BoW) ---")
count_vectorizer = CountVectorizer()
sklearn_bow = count_vectorizer.fit_transform(corpus)
print(sklearn_bow) # In ra tọa độ (hàng, cột)
print("\n")


# ==========================================
# PHẦN 3: TỰ CODE PYTHON THUẦN (MANUAL TF-IDF)
# ==========================================
# Bước 3.1: Lập từ điển độ hiếm (DF)
df_dict = {}
for w in vocab:
    df_count = sum(1 for sentence in corpus if w in sentence.split())
    df_dict[w] = df_count

# Bước 3.2: Tính trọng số TF-IDF cho từng từ trong từng câu
manual_tfidf = []
for sentence in corpus:
    words = sentence.split()
    vector = []
    for w in vocab:
        tf = words.count(w)
        df = df_dict[w]
        idf = math.log10(N / df) if df > 0 else 0 
        score = tf * idf
        vector.append(round(score, 3)) # Làm tròn 3 chữ số
    manual_tfidf.append(vector)

print("--- 3. MA TRẬN ĐẶC TỪ CODE THUẦN (TF-IDF) ---")
print(pd.DataFrame(manual_tfidf, columns=vocab))
print("\n")


# ==========================================
# PHẦN 4: DÙNG THƯ VIỆN SKLEARN (TF-IDF)
# ==========================================
print("--- 4. MA TRẬN THƯA TỪ SKLEARN (TF-IDF) ---")
tfidf_vectorizer = TfidfVectorizer()
sklearn_tfidf = tfidf_vectorizer.fit_transform(corpus)
print(sklearn_tfidf) # In ra tọa độ kèm điểm số thập phân
print("\n")

print("--- BẢNG ĐIỂM TF-IDF SKLEARN (Ép sang ma trận đặc để so sánh) ---")
# Làm tròn 3 chữ số cho giống code thuần để dễ so sánh
df_sklearn_tfidf = pd.DataFrame(sklearn_tfidf.toarray(), columns=tfidf_vectorizer.get_feature_names_out()).round(3)
print(df_sklearn_tfidf)
```
